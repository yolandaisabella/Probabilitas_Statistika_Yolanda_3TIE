```python
import pandas as pd
```

```python
#Tabel Data
# Error: FileNotFoundError. This means the file '/content/yolan.xlsx' was not found.
# Please ensure 'yolan.xlsx' is uploaded to your Colab environment.
# You can upload files by clicking the folder icon on the left sidebar, then the 'Upload to session storage' icon.
# Once uploaded, the file will typically appear in the '/content/' directory.
path = "/content/yolan.xlsx"
dataraw = pd.read_excel(path)
print(dataraw)
```

```python
#Tabel Frekuensi
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")
print(datafrq)
```

```python
#Grafik Garis
datafrq.plot()
```

```python
datafrq.plot(kind='bar')
```

```python
datafrq.plot(kind='pie', y='Frekuensi', autopct='%1.0f%%', legend=False)
```

```python
#Statistika Deskriptif
dataraw["Final Score"] = pd.to_numeric(dataraw["Final Score"], errors='coerce')
dt = dataraw["Final Score"]

stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()

print(stats)
```
