```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Tabel Data
path = "/content/yolan.xlsx"
dataraw = pd.read_excel(path)

print("DATA NILAI MAHASISWA")
print("=" * 80)
print(dataraw)
```

```python
# Tabel Frekuensi
datafrq = pd.crosstab(
    index=dataraw["Grade"],
    columns="Frekuensi"
)

datafrq["Persentase"] = (
    datafrq["Frekuensi"] / datafrq["Frekuensi"].sum() * 100
).round(2)

print("TABEL FREKUENSI GRADE")
print("=" * 40)
print(datafrq)
```

```python
# Grafik Garis
datafrq["Frekuensi"].plot(
    kind="line",
    marker="o",
    linewidth=2,
    markersize=7,
    figsize=(8, 5),
    color="#E78FB3"
)

plt.title("Distribusi Frekuensi Grade Mahasiswa")
plt.xlabel("Grade")
plt.ylabel("Jumlah Mahasiswa")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Grafik Batang
datafrq["Frekuensi"].plot(
    kind="bar",
    figsize=(8, 5),
    color="#E78FB3"
)

plt.title("Jumlah Mahasiswa Berdasarkan Grade")
plt.xlabel("Grade")
plt.ylabel("Frekuensi")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Pie Chart
colors = [
    "#F8BBD0",  # pink muda
    "#F48FB1",
    "#EC6F9E",
    "#D94F82",
    "#C2185B",
    "#880E4F"   # pink tua
]

datafrq["Frekuensi"].plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7),
    startangle=90,
    colors=colors,
    textprops={"fontsize": 11}
)

plt.title("Persentase Grade Mahasiswa")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
# Statistika Deskriptif
dataraw["Final Score"] = pd.to_numeric(
    dataraw["Final Score"],
    errors="coerce"
)

dt = dataraw["Final Score"]

stats = dt.describe()

stats["Standard Error"] = dt.sem()
stats["Median"] = dt.median()
stats["Mode"] = dt.mode().iloc[0]
stats["variance"] = dt.var()
stats["range"] = dt.max() - dt.min()
stats["skewness"] = dt.skew()
stats["kurtosis"] = dt.kurtosis()

print("STATISTIKA DESKRIPTIF FINAL SCORE")
print("=" * 50)
print(stats.round(2))
```
